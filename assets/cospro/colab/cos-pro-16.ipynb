## 문제 5. 숫자 야구 오류 수정  (10점 · 중급 · 권장 5분)

**문제 설명**

서로 다른 세 자리 숫자 비밀번호 secret과 추측 guess를 비교합니다. 숫자와 자리가 모두 같으면 스트라이크, 숫자는 있지만 자리가 다르면 볼입니다. 아래 코드에는 오류가 1개 있습니다. 찾아 수정하세요.

**매개변수 — secret, guess: 서로 다른 숫자 3개로 이루어진 길이 3의 문자열**

**return 값 — [스트라이크 수, 볼 수] 리스트를 return 하세요.**

| 입력 | return |
|---|---|
| `secret = '123', guess = '132'` | `[1, 2]` |

**예시 설명**: 1은 자리까지 같아 스트라이크 1개, 2와 3은 자리만 달라 볼 2개입니다.

<details><summary>💡 힌트</summary>

두 문자열에 공통으로 들어 있는 숫자의 수에는 스트라이크도 포함되어 있습니다.

</details>

````python
class Solution:
    def solution(self, secret, guess):
        strike = 0
        for i in range(3):
            if secret[i] == guess[i]:
                strike += 1
        common = len(set(secret) & set(guess))
        return [strike, common]


# 직접 실행해 보기 (예시)
# sol = Solution()
# ret = sol.solution(인자)
# print("solution 함수의 반환 값은", ret, "입니다.")
````

```python
        strike = 0
        for i in range(3):
            if secret[i] == guess[i]:
                strike += 1
        common = len(set(secret) & set(guess))
        return [strike, common - strike]

# 수정한 부분: return [strike, common]  →  return [strike, common - strike]
```

```python
class Solution:
    def solution(self, money):
        count = 0
```

### 문제 8. 합이 k인 연속 구간

```python
class Solution:
    def solution(self, nums, k):
        count = 0
        for i in range(len(nums)):
            total = 0
            for j in range(i, len(nums)):
                total += nums[j]
                if total == k:
                    count += 1
        return count

# 정답: k
```

### 문제 9. 공통 접두사

```python
class Solution:
    def solution(self, words):
        prefix = words[0]
        for w in words[1:]:
            while not w.startswith(prefix):
                prefix = prefix[:-1]
        return prefix
```

### 문제 10. 소수 목록

```python
class Solution:
    def solution(self, n):
        if n < 2:
            return []
        is_prime = [True] * (n + 1)
        is_prime[0] = is_prime[1] = False
        for i in range(2, int(n ** 0.5) + 1):
            if is_prime[i]:
                for j in range(i * i, n + 1, i):
                    is_prime[j] = False
        return [i for i in range(2, n + 1) if is_prime[i]]
```
