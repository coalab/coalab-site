**문제 설명**

**매개변수 — a, b: 정수 (1 이상 9999 이하, a는 b보다 크지 않음)**

**return 값 — 윤년의 개수를 return 하세요.**

| 입력 | return |
|---|---|
| `a = 2000, b = 2024` | `7` |

**예시 설명**: 2000, 2004, 2008, 2012, 2016, 2020, 2024로 7해입니다.

<details><summary>💡 힌트</summary>

'양 끝 포함'이라는 조건과 range의 끝 값 처리 방식을 비교해 보세요.

</details>

````python
class Solution:
    def solution(self, a, b):
        count = 0
        for y in range(a, b):
            if (y % 4 == 0 and y % 100 != 0) or y % 400 == 0:
                count += 1
        return count


# 직접 실행해 보기 (예시)
# sol = Solution()
# ret = sol.solution(인자)
# print("solution 함수의 반환 값은", ret, "입니다.")
````

```python
class Solution:
    def solution(self, year, month):
        days = [31, 28, 31, 30, 31, 30, 31, 31, 30, 31, 30, 31]
        leap = (year % 4 == 0 and year % 100 != 0) or year % 400 == 0
        if month == 2 and leap:
```

### 문제 10. 시각 더하기

```python
class Solution:
    def solution(self, time, add):
        h, m = map(int, time.split(":"))
        total = (h * 60 + m + add) % 1440
        return f"{total // 60:02d}:{total % 60:02d}"
```
