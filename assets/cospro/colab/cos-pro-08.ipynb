**매개변수 — n: 자연수 (1 이상 10,000 이하)**

**return 값 — 소수이면 True, 아니면 False를 return 하세요.**

| 입력 | return |
|---|---|
| `n = 7` | `True` |

**예시 설명**: 7은 1과 자기 자신만을 약수로 가지므로 소수입니다. 4, 9, 25 같은 제곱수도 정확히 걸러내야 합니다.

<details><summary>💡 힌트</summary>

range의 끝 값은 포함되지 않습니다. n의 제곱근 자신도 약수 후보입니다. 4, 9, 25로 직접 확인해 보세요.

</details>

````python
class Solution:
    def solution(self, n):
        if n < 2:
            return False
        for i in range(2, int(n ** 0.5)):
            if n % i == 0:
                return False
        return True


# 직접 실행해 보기 (예시)
# sol = Solution()
# ret = sol.solution(인자)
# print("solution 함수의 반환 값은", ret, "입니다.")
````
