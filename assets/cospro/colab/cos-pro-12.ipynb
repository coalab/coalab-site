### 문제 9. 주변 8칸의 합

```python
class Solution:
    def solution(self, matrix, r, c):
        count = 0
        for dr in (-1, 0, 1):
            for dc in (-1, 0, 1):
                if dr == 0 and dc == 0:
                    continue
                nr, nc = r + dr, c + dc
                if 0 <= nr < len(matrix) and 0 <= nc < len(matrix[0]):
                    count += matrix[nr][nc]
        return count

# 정답: len(matrix[0])
```

### 문제 10. 행렬의 곱

```python
class Solution:
    def solution(self, a, b):
        n = len(a)
        result = [[0] * n for _ in range(n)]
        for i in range(n):
            for j in range(n):
                for k in range(n):
                    result[i][j] += a[i][k] * b[k][j]
        return result
```
