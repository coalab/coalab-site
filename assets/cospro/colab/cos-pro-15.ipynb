**문제 설명**

**매개변수 — s: 올바른 괄호 문자열 (길이 2 이상 100 이하)**

**return 값 — 최대 깊이를 return 하세요.**

| 입력 | return |
|---|---|
| `s = '(()(()))'` | `3` |

**예시 설명**: 가장 안쪽에서 괄호가 3겹 겹칩니다.

<details><summary>💡 힌트</summary>

')'를 만났을 때 depth는 1만 줄어야 합니다. 현재 깊이를 어떻게 바꾸는지 확인하세요.

</details>

````python
class Solution:
    def solution(self, s):
        depth = best = 0
        for ch in s:
            if ch == "(":
                depth += 1
                best = max(best, depth)
            else:
                depth = 0
        return best


# 직접 실행해 보기 (예시)
# sol = Solution()
# ret = sol.solution(인자)
# print("solution 함수의 반환 값은", ret, "입니다.")
````

### 문제 5. 괄호의 최대 깊이 오류 수정

```python
class Solution:
    def solution(self, s):
        depth = best = 0
        for ch in s:
            if ch == "(":
                depth += 1
                best = max(best, depth)
            else:
                depth -= 1
        return best

# 수정한 부분: depth = 0  →  depth -= 1
```

### 문제 7. 요세푸스 순서

```python
class Solution:
    def solution(self, n, k):
        people = list(range(1, n + 1))
        idx = 0
        order = []
        while people:
            idx = (idx + k - 1) % len(people)
            order.append(people.pop(idx))
        return order

# 정답: len(people)
```

### 문제 8. 프린터 대기열

```python
class Solution:
    def solution(self, priorities, location):
        queue = [(p, i) for i, p in enumerate(priorities)]
        order = 0
        while queue:
            cur = queue.pop(0)
            if any(cur[0] < q[0] for q in queue):
                queue.append(cur)
            else:
                order += 1
                if cur[1] == location:
                    return order

# 정답: order
```

### 문제 10. 후위 표기식 계산

```python
class Solution:
    def solution(self, expr):
        stack = []
        for tok in expr.split():
            if tok in ("+", "-", "*"):
                b = stack.pop()
                a = stack.pop()
                if tok == "+":
                    stack.append(a + b)
                elif tok == "-":
                    stack.append(a - b)
                else:
                    stack.append(a * b)
            else:
                stack.append(int(tok))
        return stack[0]
```
