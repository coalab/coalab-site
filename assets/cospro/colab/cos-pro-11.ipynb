**매개변수 — stock: {상품: 수량} 딕셔너리 | orders: (상품, 수량) 튜플의 리스트**

**return 값 — 처리 후 재고 딕셔너리를 return 하세요.**

| 입력 | return |
|---|---|
| `stock = {'사과': 5, '배': 2}, orders = [('사과', 3), ('배', 2), ('귤', 1)]` | `{'사과': 2, '배': 0}` |

**예시 설명**: 사과 3개와 배 2개는 처리되고, 재고에 없는 귤은 무시됩니다.

<details><summary>💡 힌트</summary>

재고가 주문 수량과 정확히 같을 때도 처리해야 합니다. 비교 연산자를 확인하세요.

</details>

````python
class Solution:
    def solution(self, stock, orders):
        stock = dict(stock)
        for item, qty in orders:
            if stock.get(item, 0) > qty:
                stock[item] -= qty
        return stock


# 직접 실행해 보기 (예시)
# sol = Solution()
# ret = sol.solution(인자)
# print("solution 함수의 반환 값은", ret, "입니다.")
````

```python
# 정답: best
```

### 문제 8. 자주 나온 단어 k개

```python
class Solution:
    def solution(self, words, k):
        freq = {}
        for w in words:
            freq[w] = freq.get(w, 0) + 1
        ordered = sorted(freq, key=lambda w: (-freq[w], w))
        return ordered[:k]

# 정답: -freq[w]
```

### 문제 9. 애너그램 그룹 수

```python
class Solution:
    def solution(self, words):
        groups = set()
        for w in words:
            groups.add("".join(sorted(w)))
        return len(groups)
```

### 문제 10. 처음 중복되는 문자

```python
class Solution:
    def solution(self, s):
        seen = set()
        for ch in s:
            if ch in seen:
                return ch
            seen.add(ch)
        return ""
```
