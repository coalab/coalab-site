```python
        while i < len(a) and j < len(b):
            if a[i] <= b[j]:
                result.append(a[i])
                i += 1
            else:
                result.append(b[j])
                j += 1
        return result + a[i:] + b[j:]
```

### 문제 10. 중앙값

```python
class Solution:
    def solution(self, nums):
        ordered = sorted(nums)
        mid = len(ordered) // 2
        if len(ordered) % 2 == 1:
            return ordered[mid]
        return (ordered[mid - 1] + ordered[mid]) / 2
```
