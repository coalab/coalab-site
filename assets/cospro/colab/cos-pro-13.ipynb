**매개변수 — nums: 정수 리스트 (길이 1 이상 100 이하)**

**return 값 — 한 번 훑은 뒤의 리스트를 return 하세요.**

| 입력 | return |
|---|---|
| `nums = [5, 1, 4, 2, 8]` | `[1, 4, 2, 5, 8]` |

|---|---|

key보다 큰 앞쪽 원소를 한 칸씩 뒤로 밀 때마다 moves를 1 늘립니다.

</details>

````python
class Solution:
    def solution(self, nums):
        nums = list(nums)
        moves = 0
        for i in range(1, len(nums)):
            key = nums[i]
            j = i - 1
            while j >= 0 and nums[j] ___ key:
                nums[j + 1] = nums[j]
                j -= 1
                moves += 1
            nums[j + 1] = key
        return moves


# 직접 실행해 보기 (예시)
# sol = Solution()
# ret = sol.solution(인자)
# print("solution 함수의 반환 값은", ret, "입니다.")
````

```python
                nums[j + 1] = nums[j]
                j -= 1
                moves += 1
            nums[j + 1] = key
        return moves
```

### 문제 10. 삽입 위치 찾기

```python
class Solution:
    def solution(self, nums, target):
        low, high = 0, len(nums)
        while low < high:
            mid = (low + high) // 2
            if nums[mid] < target:
                low = mid + 1
            else:
                high = mid
        return low
```
